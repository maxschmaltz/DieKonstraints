```python
from data_utils.load_datasets import load_gecodb, load_sp18
from data_utils.gecodb_compound_parser import Compound
```

```python
get_linker_gecodb = lambda g: "_" if len((s := g.split("_"))) == 2 else s[1]

get_linker = lambda g: (
    "\"" if (l := Compound(g).linkers[0]).adds_umlaut else ""
) + l.morph
```

#### GeCoDB_v06

```python
gecodb_v06_path = "resources/comp_constr_corp_study_mar26/gecodb_v06.tsv"
gecodb_v06 = load_gecodb(gecodb_v06_path)
gecodb_v06.sample(10)
```

```python
gecodb_v06["linker_gecodb"] = gecodb_v06.index.map(get_linker_gecodb)
gecodb_v06["linker"] = gecodb_v06.index.map(get_linker)
```

```python
len(gecodb_v06)
```

```python
len(gecodb_v06["n1_lemma"].unique())
```

```python
# for L2P
# note that, even though gecodb_v06 distinguishes between linker allomorphs,
# (e.g. _+n_ vs. _+en_), the number of N1+linker is the same on practive
# no matter whether we count N1+(linker morph) or N1+(linker allomorph);
# this is explained by the fact that the allomorphs are distributed
# complementarily, and there are no cases documented in which one N1 may attach
# two allomorphs of the same linker in different compounds
gecodb_v06["n1+linker"] = gecodb_v06["n1_lemma"] + "_" + gecodb_v06["linker"]
gecodb_v06["n1+linker_gecodb"] = gecodb_v06["n1_lemma"] + "_" + gecodb_v06["linker_gecodb"]
len(gecodb_v06["n1+linker"].unique()), len(gecodb_v06["n1+linker_gecodb"].unique())
```

```python
gecodb_v06["linker"].unique()
```

```python
gecodb_v06["linker_gecodb"].unique()
```
