```python
from data_utils.load_datasets import load_gecodb, load_sp18
from data_utils.gecodb_compound_parser import Compound
```

```python
get_linker = lambda g: (
    "\"" if (l := Compound(g).linkers[0]).adds_umlaut else ""
) + l.morph

get_linker_gecodb = lambda g: "_" if len((s := g.split("_"))) == 2 else s[1]
```
