```python
import re

from data_utils.load_datasets import load_gecodb, load_sp18
```

#### GeCoDB_v06

```python
gecodb_v06_path = "resources/comp_constr_corp_study_mar26/gecodb_v06.tsv"
gecodb_v06 = load_gecodb(gecodb_v06_path)
gecodb_v06.sample(10)
```

```python
len(gecodb_v06)
```

```python
len(gecodb_v06["n1_lemma"].unique())
```

```python
# for L2P
n1_linker = lambda g: re.sub(r"_[^_]+$", "", g)
gecodb_v06["n1+linker"] = gecodb_v06.index.map(n1_linker)
len(gecodb_v06["n1+linker"].unique())
```

```python
linker = lambda n1l: "_" if not "_" in n1l else n1l.split("_")[1]
gecodb_v06["linker"] = gecodb_v06["n1+linker"].map(linker)
gecodb_v06["linker"].unique()
```
