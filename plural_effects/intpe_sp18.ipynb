# Internal Plural Effect

This notebook investigates the internal plural effect (IntPE) in German compounding as defined by Schäfer & Pankratz 2018:

> Certain semantic classes of N2s standing in an appropriate semantic relation with N1 might force N1 to have a plural interpretation and therefore lead to a preference for using the pluralic linking element. We call this the internal plural effect.


## Background

Schäfer & Pankratz 2018 assembled a German compound dataset of N=9415 compounds built from 48 first constituent (N1s), up to ten N1 for each of the seven pluralic linkers (PLLs) -_n_-, -_en_-, -_e_-, -_"e_-, -_"_-, -_er_-, -_"er_-. All N1s are countable nouns which are productively used in compounds both with and without the pluralic linking element. For each N1, up to 100 compound **tokens** with a non-pluralic (mostly zero or -_s_-) and up to 100 compound **tokens** with the pluralic linker were retrieved from DECOW16A. Since tokens and not types were counted, there is a number of duplicated constituent combinations (CC) in the data, though sometimes with alternating linkers. Besides, the original wordforms were preserved, and so compounds both in their Sg and Pl form appear, sometimes within the duplicated CC. Thus for example, the CC 'Apfel' + 'Baum' features occurrences 'Apfelbaum', 'Apfelbäume', 'Äpfelbaum', and 'Äpfelbäume' in the data. Note that no frequency threshold on the compounds was set while sampling the data.

What Schäfer & Pankratz 2018 were interested in regarding the IntPE was whether the probability of the pluralic linker increases in compounds in which the second constituent (N2) forces a plural meaning of N1. In other words, whether there is an association between the plural semantics of N1 and the pluralic linker. While annotating the Pl meaning of N1s, they found a number of uncertain examples, so they made two measurements: a strict measurement (the uncertain cases were treated as a Sg N1 meaning) and a lax measurement (the uncertain cases were treated as a Pl N1 meaning). The experiment revealed that the difference between the two measurements is negligible (see Schäfer & Pankratz 2018, 29), and the lax measurement brings more positive result. We therefore only provide the results of the strict version of the original study below, as well as conduct our own experiments in the strict paradigm.

To investigate the IntPE, Schäfer & Pankratz 2018 utilized the following methodology.

1. First, they created a 2x2 contingency table over the data with the following axes: non-pluralic vs. pluralic linker, and Sg vs. Pl N1 meaning. The resulting table is below:

	|  | Pl | Sg |
	|---|---|---|
	| non-pluralic linker | 214 | 4493 |
	| pluralic linker | 697 | 4010 |

2. Second, they applied the Chi2 test to identify whether there is a relation between the two variables (the "plurality" of the linker vs. the formal number of the compound). The test produced a significant result at sig = 0.05 $(χ2 = 282.343, df = 1, p ~= 0.0)$.

3. Finally, they measured the effect size with Cramer's V and revealed that it is not very high at $v = 0.173$.

<small>
Note that, as opposed to the ExtPE, they do not provide the symmetrical (standardised) residuals values, but our own (legacy) experiments resulted in a strong (~13-17) positive trend between the Pl meaning and PLLs.
</small>

Schäfer & Pankratz 2018 conclude that the IntPE is confirmed.

As an additional confirmation, the preceding Pankratz 2017 thesis was based on the same methodology and gave similar significant results for the IntPe on a different dataset.


## Objective

When scrutinizing the dataset from Schäfer & Pankratz 2018, a few groups of compounds could be detected that could decrease the overall effect size:
1. Compounds whose N1 only appear with a Sg or only with a Pl meaning. Since the IntPE is triggered by semantical plurality of the N1, this N1 should occur both with a Sg and with a Pl meaning in order to validate whether the IntPE takes place. Otherwise, it is impossible to make judgments about the expected behavior of these compounds with the opposite meaning.
2. Compounds whose N1s are forced to attach the pluralic linker  due to other constraints, e.g. because of schwa ('Katzenfell' vs. 'Katzenfelle', found in the data) or as a person designation ('Kinderbuch' vs. 'Kinderbücher', found in the data). If the given N1 mostly takes a pluralic linker, it will most likely do with a Pl meaning of N1, making the IntPE undetectable.

**Hypothesis 1** was the following: _the IntPE is present regardless of "data contamination" and frequency setting_.

**Hypothesis 2** focuses on the effect of the "problematic" groups on ExtPE: _after removing these "problematic" groups from the data, the IntPE should be more pronounced in the remaining compounds; gradually letting the problematic groups back in should weaken the IntPE_.

Since the IntPE was confirmed even on the original dataset with no frequency filtering, we proposed **Hypothesis 3**: _the IntPE should be agnostic to the frequency setting_.


## Data

<small>
Convention: <em>(N=x)</em> refers to the number of compound <b>tokens</b> in the dataset, <em>(UniqN=x)</em> refers to the number of unique compound <b>lemmata</b> in the dataset, <em>(UniqN1N=x)</em> refers to the number of unique N1 <b>lemmata</b> in the dataset, "compounds" refers to "compound tokens" unless specified otherwise.
</small>

In addition to the original dataset, we prepared two modifications of the original Schäfer & Pankratz 2018 dataset resulting after removing the problematic groups gradually.
* `alt` ( **ALT**ernating): compounds built from N1s appearing both with a Sg and with a Pl meaning | $(N=8025, UniqN=6746, UniqN1N=45)$.
* `alt_npll` (**ALT**ernating, **N**o **PL**uralic **L**inker allowed): Compounds built from N1s appearing both with a Sg and with a Pl meaning that may not attach a pluralic linker | $(N=5311, UniqN=4449, UniqN1N=30)$.

In parallel, we defined three levels of frequency (the word counts (WCs) were obtained from DeReKo following the procedure outlined in Shmalts 2026, accessed mid June 2026):
* `lfreq` (low frequency): word count (WC) $> 0$, as in the original data.
* `mfreq` (moderate frequency): WC $>= 20$.
* `hfreq` (high frequency): WC $>= 100$.

During the studies, each of the three datasets was dynamically filtered to conform to the three frequency levels. Thus, nine unique data samples appear in the studies.


## Workflow

To evaluate the proposed hypothesis, we run three studies.

#### Study 1

Study 1 replicates the original IntPE study from Schäfer & Pankratz 2018 on the six data samples (two dataset modifications x three frequency levels). It validates whether the IntPE is present regardless of "data contamination and frequency setting". Comparing the individual results from the samples will additionally give an overall idea of the impact of the problematic groups and the frequency level on IntPE's strengths.

Study 1 contributes to **Hypothesis 1**. Expected outcome: the IntPE will be significant in all the data samples.

It also indirectly contributes to **Hypothesis 2** and **Hypothesis 3**. Expected outcome: the increase in the level of "contamination" will result in a weaker effect size. The increase of frequency will not affect the effect size.

### Study 2

Study 2 focuses on the distribution of non-pluralic and pluralic linker in compounds with a Sg meaning only. It contrasts the distribution of PLL and NLL in the original dataset vs each of the two modifications.

Study 2 contributes to **Hypothesis 2** and provides a more detailed account of the effect of the problematic groups on IntPE. Expected outcome: more "contaminated" datasets will have a larger share of pluralic linkers in compounds with a Sg meaning (due to more possible sources).

### Study 3

Study 3 focuses on the distribution of non-pluralic and pluralic linker in compounds with a Pl meaning only. It contrasts the distribution in the low frequency setting vs the moderate/high frequency setting on the `alt` dataset (see explanation in Results below).

Study 3 contributes to **Hypothesis 3** and provides a more detailed account of the effect of the frequency setting on IntPE. Expected outcome: the share of the pluralic linkers in compounds with a Pl meaning will be the same in the low frequency setting as in other settings.


## Results

### Summary

The findings from the three studies **confirm Hypothesis 1** that the IntPE is significant regardless of the "data contamination" and frequency setting. More specifically, both "problematic" group and frequency effects are not significant, which **refutes Hypothesis 2** and **supports Hypothesis 3**.

# Study 1

Study 1 replicates the original IntPE study from Schäfer & Pankratz 2018 on the six data samples (two dataset modifications x three frequency levels). It validates whether the IntPE is present regardless of "data contamination and frequency setting". Comparing the individual results from the samples will additionally give an overall idea of the impact of the problematic groups and the frequency level on IntPE's strengths.

Study 1 contributes to **Hypothesis 1**. Expected outcome: the IntPE will be significant in all the data samples.

It also indirectly contributes to **Hypothesis 2** and **Hypothesis 3**. Expected outcome: the increase in the level of "contamination" will result in a weaker effect size. The increase of frequency will not affect the effect size.

```python
label_pfx = "n1_"
label_1, label_2 = "coll", "ind"

def lrun_onesample_study(
	sample: pd.DataFrame,
	outdir: Optional[str]=None
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.Series]:
	return run_onesample_study(
		sample,
		groupby=groupby,
		target_col=target_col,
		label_pfx=label_pfx,
		label_1=label_1,
		label_2=label_2,
		outdir=outdir
	)
```

```python
st1dir = os.path.join(resdir, "study_1")
os.makedirs(st1dir, exist_ok=True)
```

```python
chi2_summaries = []
with tqdm(
    total=len(datamods) * len(freq_levels), 
    desc="IntPE Study 1"
) as pbar:

    # two dataset modifications x ...
    for sample_sfx, sp18_sample in datamods:

        chi2dfs = []
        outdir = os.path.join(st1dir, sample_sfx)

        # ... x three levels of frequency
        for freq_sfx, freq_threshold in freq_levels:

            sp18_subsample = sp18_sample[
                sp18_sample["wc"] >= freq_threshold
            ]
            *_, chi2df = lrun_onesample_study(
                sp18_subsample,
                os.path.join(outdir, freq_sfx)
            )
            # will become the colname in chi2_summary
            chi2df.name = freq_sfx
            chi2dfs.append(chi2df)
            
            pbar.update(1)

        # summary over the freq levels for one dataset
        chi2_summary = pd.concat(chi2dfs, axis=1)
        savetsv(chi2_summary, outdir, "chi2_summary")
        chi2_summaries.append(chi2_summary)
```

The following table provides an overview over the results of Study 1. For each combination of the dataset and the frequency setting, the results of the Chi2 test over the resulting sample is given. We can see that none of the p-values are even near low.

```python
# 3D res DF: dataset x freq x chi2 test metrics
chi2_summary = pd.concat(chi2_summaries, keys=[p[0] for p in datamods])  \
    .swaplevel(0, 1)    \
    .sort_index(level=0, sort_remaining=False)  \
    .reindex(chi2df.index, level=0)
savetsv(chi2_summary, st1dir, "chi2_summary")
chi2_summary
```

We can observe the p-values consistently approximate $0.0$ for all the data samples. There is no observable difference between the Cramer's V values across the datasets whatsoever. However, we can see a consistent increase of these values from lower to higher frequency.

#### Appendix to Study 1

Other that the analysis from Schäfer & Pankratz, we had an idea of performing a paired t-test between the shares of the pluralic linker in Sg and Pl forms. The idea was to interpret the two samples as measurements from the same population (N1s) taken before and after a treatment, "semantic pluralization" playing the role of treatment. The Shapiro test revealed that the PLL share in compounds with a Sg and with a Pl meaning are distributed normally at sig = 0.05 (Sg meaning: $W = 0.93, p = 0.05$; Pl meaning: $W = 0.934, p = 0.062$) in the "cleanest" `alt_npll` dataset (but not in `alt`). The normality of both distributions and the sample size of $30$ satisfy the assumptions of the t-test. The t-test results (alternative: less in Sg) additionally confirm the IntPE ($t = -8.237, df = 29, p ~= 0.0$).

```python
label_1_calc_shares, label_2_calc_shares = label_1, label_2

def lrun_twosamples_study(
	sample_1: pd.DataFrame,
	sample_2: pd.DataFrame,
	label_1: str,
	label_2: str,
	target_predictor: str,
	outdir: Optional[str]=None
):
	return run_twosamples_study(
		sample_1,
		sample_2,
		groupby=groupby,
		target_col=target_col,
		label_1_calc_shares=label_1_calc_shares,
		label_2_calc_shares=label_2_calc_shares,
		label_1=label_1,
		label_2=label_2,
		target_predictor=target_predictor,
		outdir=outdir
	)
```

```python
st2dir = os.path.join(resdir, "study_2")
os.makedirs(st2dir, exist_ok=True)
```

```python
# problematic groups effect; compare the original dataset
# with the two modifications
with tqdm(total=len(datamods), desc="IntPE Study 2") as pbar:

    chi2dfs = []

    # two dataset pairings
    for (sample_sfxm, sp18_sample_1) in datamods:

        # we expect a larger pluralic linker share in 
        # the more "contaminated" sample in Sg form because
        # they should have more sources for the pluralic linker
        # to appear in the Sg form
        *_, chi2df = lrun_twosamples_study(
            sp18_sample_1,
            sp18,
            label_1=sample_sfx,
            label_2="sp18",
            # look at the distr with a Sg meaning
            target_predictor="ind",
            outdir=os.path.join(st2dir, f"vs_{sample_sfx}")
        )
        # will become the colname in chi2_summary
        chi2df.name = f"vs_{sample_sfx}"
        chi2dfs.append(chi2df)
        
        pbar.update(1)

    chi2_summary = pd.concat(chi2dfs, axis=1)
    savetsv(chi2_summary, st2dir, "chi2_summary")
```

### The N1 IntPE Study.

```python
n1s = alt["n1"].unique()
n1s
```

```python
chi2_summaries = defaultdict(dict)
sign_n1s = []
with tqdm(
    total=len(datamods) * len(freq_levels) * len(n1s), 
    desc="N1 IntPE Study"
) as pbar:

    # three datasets x ...
    for (sample_sfx, sp18_sample) in datamods:
        
        # ... x 45 N1s x ...
        for n1 in n1s:

            chi2dfs = []
            is_sign = False

            # ... x three levels of frequency
            for freq_sfx, freq_threshold in freq_levels:

                sp18_subsample = sp18_sample[
                    (sp18_sample["wc"] >= freq_threshold) &
                    (sp18_sample["n1"] == n1)
                ]

                m = len(sp18_subsample["n1"].unique())

                try:
                    # not providing the directory; only 
                    # significant N1s will be saved later
                    *_, chi2df = lrun_onesample_study(sp18_subsample)
                    # increased original p-value to account for small sample sizes;
                    # however, when a large family of significance tests is run
                    # simultaneously, there is a probability that some
                    # significant results will appear by chance; to account for that,
                    # we follow [schaefer_pankratz_2018:27] and apply Šidák’s
                    # correction; the corresponding R code is:
                    # adjust.p.sidak <- function(p, m) { 1-(1-p)^m };
                    chi2df["p_sidak"] = (1 - (1 - chi2df["p"]) ** m).round(3)
                    # will become the colname in chi2_summary
                    chi2df.name = freq_sfx
                    chi2dfs.append(chi2df)
                    # use the corrected p-value!
                    if chi2df["p_sidak"] < 0.05:
                        sign_n1s.append(n1)
                except (KeyError, ValueError):
                    # `all_nipe_npll` is missing a few N1s;
                    # otherwise, after freq filtering
                    # not all CCs have valid contingency tables
                    pass

                pbar.update(1)
            
            # summary over the freq levels for one N1 in one dataset
            chi2_summary = pd.concat(chi2dfs, axis=1) if chi2dfs else None
            chi2_summaries[n1][sample_sfx] = chi2_summary
```

Twenty-five N1s reached a statistical significance at sig = 0.05 in the low frequency setting. 13 did not reach a significance level at sig = 0.05 on the `alt_npll` dataset, and 20 on the `alt` dataset.

```python
# frequency effect; compare `lfreq` to
# `mfreq`/`hreq` in each dataset
with tqdm(
    total=len(datasets) * (len(freq_levels) - 1),
    desc="Extenden Freq Study "
) as pbar:

    chi2_summaries = []

    # three datasets x ...
    for (sample_sfx, sp18_sample_1) in datasets:

        chi2dfs = []

        # ... x two frequency level parings
        for freq_sfx, freq_threshold in freq_levels[1:]:

            sp18_sample_2 = sp18_sample_1[
                sp18_sample_1["wc"] >= freq_threshold
            ]

            try:
                # we expect a larger pluralic linker share
                # in the `lfreq` sample
                *_, chi2df = lrun_twosamples_study(
                    sp18_sample_1,
                    sp18_sample_2,
                    label_1="lfreq",
                    label_2=freq_sfx,
                    # Pl meaning only
                    target_predictor="coll",
                    outdir=os.path.join(outdir, f"vs_{freq_sfx}")
                )
                # will become the colname in chi2_summary
                chi2df.name = f"vs_{freq_sfx}"
                chi2dfs.append(chi2df)
            except (KeyError, ValueError):
                # `all_nipe_npll` is missing a few N1s;
                # otherwise, after freq filtering
                # not all CCs have valid contingency tables
                pass

            pbar.update(1)

        # summary over the freq level pairings for one dataset
        chi2_summary = pd.concat(chi2dfs, axis=1) if chi2dfs else None
        if chi2_summary is not None:
            savetsv(chi2_summary, outdir, "chi2_summary")
        chi2_summaries.append(chi2_summary)
```
