```python
import os
import pandas as pd
from collections import defaultdict
```

```python
sp18_ext_path = "../resources/developed/sp18/"

resdir = "../out/internal_effect"

incsv_args = {
	"sep": "\t",
    "header": 0,
    "index_col": "gecodb",
    # all str, WC int, is linker pluralic bool
    "dtype": defaultdict(
        lambda: str,
        wc="int",
        is_linker_pl="bool"
    ),
	# = csv.QUOTE_NONE, separate by tabs only
    "quoting": 3
}

outcsv_args = {
    "sep": "\t",
    "index": True,
    "header": True
}
```

```python
def read_sp18(name: str) -> pd.DataFrame:
	return pd.read_csv(
		os.path.join(sp18_ext_path, f"{name}.tsv"),
		**incsv_args
	)
```

```python
sp18 = read_sp18("sp18")
sp18.sample(10)
```

```python
sp18 = sp18[~sp18.index.duplicated(keep="first")]
sp18
```

```python
sp18 = sp18[sp18["wc"] >= 20]
sp18
```

```python
# coverage
def intpe_is_applicable(row: pd.Series, strict=True) -> bool:
    cond = ["coll"] if strict else ["coll", "konvc"]
    return row["n1_sem"] in cond

# regularity
def intpe_applies(row: pd.Series) -> bool:
    return row["is_linker_pl"]

def comp_constraints(row: pd.Series):
    to_return = []
    for s in [True, False]:
        is_applicable = intpe_is_applicable(row, strict=s)
        applies = intpe_applies(row) if is_applicable else False
        to_return.extend([is_applicable, applies])
    return to_return
```

```python
sp18["is_appl_strict"], sp18["applies_strict"], sp18["is_appl_lax"], sp18["applies_lax"] = zip(
    *sp18.apply(comp_constraints, axis=1)
)
```

```python
N = len(sp18)
intpe = pd.DataFrame({
    "strict": {
        "cvg": (xs := len(
            sp18[sp18["is_appl_strict"]]
        )) / N,
        "reg": len(
            sp18[sp18["applies_strict"]]
        ) / xs
    },
    "lax": {
        "cvg": (xl := len(
            sp18[sp18["is_appl_lax"]]
        )) / N,
        "reg": len(
            sp18[sp18["applies_lax"]]
        ) / xl
    }
})

intpe
```
