```python
# p2l:decl_cl:pl:#e-0|pl_interpr-e
#
# First constituents that are constituted by nouns that build
# the plural form with -e attach -e- irregularly in compounds
# in which the second constituent forces
# a plural meaning of the given first constituent.
def plur_e_pl_is_applicable(row: pd.Series, strict=True) -> bool:
	cond = ["coll"] if strict else ["coll", "konvc"]
	return (
		# morph cond
		(row["n1_pl_marker"] == "_+e_") and
        # sem cond
        row["n1_sem"] in cond
    )

def plur_e_pl_applies(row: pd.Series, strict=True) -> bool:
    return row["is_linker_pl"]


# p2l:decl_cl:pl:#er-0/er|!pl_interpr-0
#
# First constituents that are constituted by nouns that build
# the plural form with -(")er
# attach a zero linker in majority of compounds
# in which the second constituent forces
# a singular meaning of the given first constituent.
def plur_er_sg_is_applicable(row: pd.Series, strict=True) -> bool:
	return (
		# morph cond
		(row["n1_pl_marker"] in ["_+er_", "_+=er_"]) and
        # sem cond
        row["n1_sem"] == "ind"
    )

def plur_er_sg_applies(row: pd.Series, strict=True) -> bool:
    return not row["is_linker_pl"]


# p2l:decl_cl:pl:#er-0/er|pl_interpr-er
#
# First constituents that are constituted by nouns that build
# the plural form with -(")er
# attach -(")er- more likely in compounds
# in which the second constituent forces
# a plural meaning of the given first constituent.
def plur_er_pl_is_applicable(row: pd.Series, strict=True) -> bool:
	cond = ["coll"] if strict else ["coll", "konvc"]
	return (
		# morph cond
		(row["n1_pl_marker"] in ["_+er_", "_+=er_"]) and
        # sem cond
        row["n1_sem"] in cond
    )

def plur_er_pl_applies(row: pd.Series, strict=True) -> bool:
    return row["is_linker_pl"]


# p2l:decl_cl:pl:#e_uml-0|pl_interpr-e_uml
#
# First constituents that are constituted by nouns that build
# the plural form with -e and umlaut attach -"e- irregularly in compounds
# in which the second constituent forces
# a plural meaning of the given first constituent.
def plur_e_uml_pl_is_applicable(row: pd.Series, strict=True) -> bool:
	cond = ["coll"] if strict else ["coll", "konvc"]
	return (
		# morph cond
		(row["n1_pl_marker"] == "_+=e_") and
        # sem cond
        row["n1_sem"] in cond
    )

def plur_e_uml_pl_applies(row: pd.Series, strict=True) -> bool:
    return row["is_linker_pl"]


# p2l:decl_cl:pl:#0_uml-0|pl_interpr-0_uml
#
# First constituents that are constituted by nouns that build
# the plural form with a zero ending and umlaut attach
# a zero linker with umlaut irregularly in compounds
# in which the second constituent forces
# a plural meaning of the given first constituent.
def plur_0_uml_pl_is_applicable(row: pd.Series, strict=True) -> bool:
	cond = ["coll"] if strict else ["coll", "konvc"]
	return (
		# morph cond
		(row["n1_pl_marker"] == "_+=_") and
        # sem cond
        row["n1_sem"] in cond
    )

def plur_0_uml_pl_applies(row: pd.Series, strict=True) -> bool:
    return row["is_linker_pl"]



# p2l:tend:pl_interpr-pl_marker|def-pl_marker
#
# Linkers that are identical to the plural ending
# of the first constituent in a given compound
# tend to be associated with
# a plural meaning of this first constituent within the compound.
def pl_interpr_is_applicable(row: pd.Series, strict=True) -> bool:
	return row["is_linker_pl"]

def pl_interpr_applies(row: pd.Series, strict=True) -> bool:
	cond = ["coll"] if strict else ["coll", "konvc"]
	return row["n1_sem"] in cond
```

```python
n = len(sp18)
stats = {}
for id, func_is_a, func_a in [
    ("plur_e_pl", plur_e_pl_is_applicable, plur_e_pl_applies),
    ("plur_er_sg", plur_er_sg_is_applicable, plur_er_sg_applies),
    ("plur_er_pl", plur_er_pl_is_applicable, plur_er_pl_applies),
    ("plur_e_uml_pl", plur_e_uml_pl_is_applicable, plur_e_uml_pl_applies),
    ("plur_0_uml_pl", plur_0_uml_pl_is_applicable, plur_0_uml_pl_applies),
    ("pl_interpr", pl_interpr_is_applicable, pl_interpr_applies)
]:
    for strict in [True, False]:
        apply = lambda row: (
            (is_applicable := func_is_a(row, strict)),
            func_a(row, strict) if is_applicable else False
        )
        strict_str = "strict" if strict else "lax"
        sp18[f"{id}_is_applicable_{strict_str}"], sp18[f"{id}_applies_{strict_str}"] = zip(
            *sp18.apply(apply, axis=1)
        )
        cvg_abs = len(sp18[sp18[f"{id}_is_applicable_{strict_str}"]])
        reg_abs = len(sp18[sp18[f"{id}_applies_{strict_str}"]])
        stats[(id, strict_str)] = {
            "cvg": cvg_abs / n,
            "reg": reg_abs / cvg_abs
        }
```

```python
stats_df = pd.DataFrame(stats).T.sort_index(ascending=False)
stats_df
```
